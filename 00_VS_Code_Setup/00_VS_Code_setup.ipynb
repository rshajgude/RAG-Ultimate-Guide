6. **Install packages whenewer required** and keep track in `requirements.txt` with its version.Also it will be good in learning if you keep `requirements.txt` empty and add as you need. 
   

## 1. Clone repository for channel playlist - 

**Run below clone command in your terminal/cmd where you want to clone repo**

```bash
git clone https://github.com/rshajgude/RAG-Ultimate-Guide.git
```

## 2. Creating a virtual environment

**Run below command to create a virtual environment named "env" in VS terminal.**

```bash
uv venv env --python 3.12
```


Remember, **"env"** is name of environment which will be created and **"uv venv"** these are keywords. 

And **"--python 3.12"** is telling you virtual environment is python based with python version 3.12

## 3. Activate virtual environment

**Run below command to activate environment "env". (If you used another name change below path)**

```bash
source env/bin/activate
```

## 4. install ipykernel

**Use command -**

```bash
uv pip install ipykernel
```
